```python
# Count how many books have average rating > 4
count = (df['average_rating'] > 4).sum()

print(f"Number of books with average rating > 4: {count}")
```

```python
# Count books published in 2010
count_2010 = (df['year'] == 2010).sum()

print(f"Number of books published in 2010: {count_2010}")
```

```python
# Filter for DAW publications with average_rating > 4
authors = df[(df['publisher'] == 'DAW') & (df['average_rating'] > 4)]['author'].unique()

print("Authors with DAW publications and average_rating > 4:")
print(authors)
```

```python
# Count books with more than 500 pages published after 2010
count = df[(df['num_pages'] > 500) & (df['year'] > 2010)].shape[0]

print(f"Number of books with more than 500 pages published after 2010: {count}")
```

```python
# Count how many books are authored by Agatha Christie
count = (df['author'] == 'Agatha Christie').sum()

print(f"Number of books written by Agatha Christie: {count}")
df['author'].value_counts().head(10)
```

```python
# Count books published in Spanish after 2000
count_spa = df[(df['language_code'] == 'spa') & (df['year'] > 2000)].shape[0]

print(f"Number of books published in Spanish after 2000: {count_spa}")

df_spa = df[df['language_code'] == 'spa']
df_spa_after_2000 = df_spa[df_spa['year'] > 2000]
df_spa_after_2000.shape
```
