```python
# Importing necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Data sanity checks
import ydata_profiling

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import recall_score
from sklearn.metrics import confusion_matrix
```

```python
df = pd.read_csv("backpain.csv")
df.head()
```

```python
ydata_profiling.ProfileReport(df)
```

```python
# encode Status (Abnormal -> 1, Normal -> 0)
df["Status"] = df["Status"].astype(str).str.strip().str.lower()

# encode: abnormal -> 1, normal -> 0
df["Status"] = (df["Status"] == "abnormal").astype(int)

df.head()
```

```python
# features & target
X = df.drop("Status", axis=1)
y = df["Status"]

# split WITHOUT stratify
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=1
)
```

```python
# Fit the model on training set
model = LogisticRegression(random_state=1)
lg = model.fit(X_train,y_train)

# predicting on the training set
y_train_pred = lg.predict(X_train)
```

```python
recall_score(y_train,y_train_pred)
```

```python
# logistic regression (default params, random_state=1)
log_reg = LogisticRegression(random_state=1, max_iter=1000)
log_reg.fit(X_train, y_train)

# predictions on training set
y_train_pred = log_reg.predict(X_train)

# recall for class 1 (Abnormal)
train_recall = recall_score(y_train, y_train_pred, pos_label=1)

print("Training Recall:", train_recall)
```

```python
cm = confusion_matrix(y_train, y_train_pred)
cm
```

```python
# logistic regression (default params, random_state=1)
log_reg = LogisticRegression (solver='liblinear', random_state=1)
log_reg.fit(X_train, y_train)

# predictions on training set
y_train_pred = log_reg.predict(X_train)

# recall for class 1 (Abnormal)
train_recall = recall_score(y_train, y_train_pred, pos_label=1)

print("Training Recall:", train_recall)
```

```python
# predictions on TEST set
y_test_pred = log_reg.predict(X_test)

# confusion matrix
cm = confusion_matrix(y_test, y_test_pred)
print(cm)

# extract FN (row 1, col 0)
FN = cm[1, 0]
print("False Negatives:", FN)
```
